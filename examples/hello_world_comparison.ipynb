# Hello World: Atlas Cortex vs LangChain

Stop destroying your RAG context with character chunking. This notebook demonstrates how **Atlas Cortex** uses an AST-based semantic router (built in Rust) to save tokens and improve retrieval accuracy by preserving the connective tissue of your documents.

```python
import os
import sys

# Configurando PYTHONPATH para carregar o módulo nativo do Atlas Cortex
base_dir = os.path.dirname(os.path.abspath(''))
if base_dir not in sys.path:
    sys.path.append(base_dir)

from python.atlas_cortex.integrations.langchain import AtlasCortexSplitter
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

### 1. Load a Complex Markdown Document
We'll load a sample markdown file (`sample_complex_doc.md`) that contains tables, nested headings, and code blocks.

```python
file_path = "sample_complex_doc.md"
with open(file_path, "r", encoding="utf-8") as f:
    markdown_text = f.read()

print(f"Document Loaded: {len(markdown_text)} characters.")
```

### 2. The Old Way: LangChain RecursiveCharacterTextSplitter
Using mechanical constraints (`chunk_size=300`) forces overlapping and destroys contextual borders.

```python
langchain_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50,
    length_function=len
)

langchain_chunks = langchain_splitter.split_text(markdown_text)

print(f"LangChain generated {len(langchain_chunks)} fragmented chunks.\n")
print("Example of a bad chunk (context lost):")
print("-" * 40)
print(langchain_chunks[3])
print("-" * 40)
```

### 3. The Atlas Cortex Way (AST Semantic Routing)
Extracts nodes based on the semantic Abstract Syntax Tree (AST), ensuring code blocks, tables, and paragraphs remain connected to their parents.

```python
atlas_splitter = AtlasCortexSplitter()
atlas_chunks = atlas_splitter.split_text(markdown_text)

print(f"Atlas Cortex generated {len(atlas_chunks)} atomic semantic nodes.\n")
print("Example of a perfect semantic node:")
print("-" * 40)
print(atlas_chunks[1])
print("-" * 40)
```

### 4. Conclusion
By migrating from `RecursiveCharacterTextSplitter` to `AtlasCortexSplitter`, you drop 1 line of code into your LangChain pipeline and get **Enterprise-grade AST parsing**, zero overlap entropy, and up to 60% savings on context window limits.
